Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.

đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.

```python
arr = list(map(int, input().split()))
new_arr = set(arr)
k = 0
y = 0

for i in new_arr:
    k = arr.count(i)
    if k < i:
        y += k
    if k > i:
        y += k - i
    if k == i:
        y += 0
print(y)
        
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ 
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. 
Nếu không có phần tử nào thỏa mãn thì không in gì 

```python
# Khong dung Boyer-Moore va dung luong bo nho la O(n^2)

def Xuat_hien(arr):
    n = int(len(arr))
    for i in arr:
        k = arr.count(i)
        if k > n//2:
            return i
    return None
arr = eval(input())
print(Xuat_hien(arr))
```

```python
def BoyerMoore(arr):
    if len(arr) == 0:
        return None
    
    candidate = None
    count = 0

    for i in arr:
        if count == 0:
            candidate = i
            count = 1
        elif i == candidate:
            count += 1
        else:
            count -= 1

    if arr.count(candidate) > len(arr) // 2:
        return candidate
    return None

arr = eval(input())
print(BoyerMoore(arr))
```

Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần; 

```python
def sort_codition(arr):
    even = []
    odd = []
    for i in arr:
        if i % 2 == 0:
            even.append(i)
        else:
            odd.append(i)

    even = sorted(even)

    for j in range(len(odd)-1):
        for k in range(j+1, len(odd)):
            if odd[j] < odd[k]:
                odd[j], odd[k] = odd[k], odd[j]
                
    return even + odd

arr = eval(input())
print(sort_codition(arr))

```

W7A21
Cho N là ngày, T là tháng năm 2020, Tính xem đó là ngày thứ bao nhiêu
của năm đó nếu đó là ngảy hợp lệ

```python
N = int(input())
T = int(input())

def day_in_year(N, T):
    if T < 1 or T > 12:
        return "Invalid month"
    
    match T:
        case 1|3|5|7|8|10|12:
            max_day = 31
        case 4|6|9|11:
            max_day = 30
        case 2:
            max_day = 29

    if N < 1 or N > max_day:
        return "Invalid day"
    day_of_year = sum([31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30][:T-1]) + N
    return day_of_year
print(day_in_year(N, T))
```

Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
n = int(input())
full_name = [list(map(str, input().split())) for _ in range(n)]
name = [i[-1] for i in full_name]
max_count = 0
max_name = ""

for i in name:
    k = name.count(i)
    if k > max_count:
        max_count = k
        max_name = i
    elif k == max_count:
        if i < max_name:
            max_name = i
print(max_name)
```

Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số

```python
n = input().strip()
k = int(input())
rem = len(n) - k
res = []

for digit in n:
    while res and res[-1] < digit and rem > 0:
        res.pop()  
        rem -= 1   
    res.append(digit)

print("".join(res[:k]))
```

Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)

```python
def prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def consencutive_num(lst):
    if not lst:
        return 0
    consecutive = [lst[0]]
    max_length = [lst[0]]

    for i in range(1, len(lst)):
        if lst[i] == lst[i-1] + 1:
            consecutive.append(lst[i])
        else:
            if len(consecutive) > len(max_length):
                max_length = consecutive
            consecutive = [lst[i]]
    
    if len(consecutive) > len(max_length):
        max_length = consecutive
    
    return max_length

n = int(input())
res = []

for i in range(2, n):
    if not prime(i):
        res.append(i)

print(*consencutive_num(res))
```

Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
def consencutive(lst):
    if not lst:
        return False
    
    consencutive = [lst[0]]
    max_len = [lst[0]]
    
    for i in range(1, len(lst)):
        if lst[i] > lst[i-1]:
            consencutive.append(lst[i])
        else:
            if len(consencutive) > len(max_len):
                max_len = consencutive
            consencutive = [lst[i]]

    if len(consencutive) > len(max_len):
        max_len = consencutive
    
    return max_len

lst = list(map(int, input().split()))
print(consencutive(lst))
```
