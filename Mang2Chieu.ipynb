Bài 2A: Tính tổng cột k của ma trận nxm số nguyên (n, m vào từ bàn phím)

```python
n, m = map(int, input().split())
k = int(input())
matrix = [list(map(int, input().split())) for _ in range(n)]
sum_col = 0

for i in range(n):
    sum_col += matrix[i][k-1]

for i in range(n):
    for j in range(m):
        print(matrix[i][j], end=' ')
    print()
print(sum_col)
```

Bài 4: Trả về hàng (chỉ số bắt đầu từ 0) có tổng hàng nhỏ nhất trong ma trận n x m. 
Viết hàm row_with_min_sum(mat) trả về chỉ số bắt đầu từ 0 của hàng có tổng nhỏ nhất. 
Nếu có nhiều hàng cùng tổng nhỏ nhất, trả về hàng có chỉ số lớn nhất. - Dùng vòng lặp để tính tổng từng hàng. -. 
n = 2
m = 3
mat = [[1, 3, 5], 
       [6, 2, -9]]


```python
n, m = map(int, input().split())
matrix = [list(map(int, input().split())) for _ in range(n)]


def row_with_min_sum(matrix):
    min_sum = float('inf')
    list_index = []

    for i in range(len(matrix)):
        row_sum = sum(matrix[i])
        if row_sum <= min_sum:
            min_sum = row_sum
            list_index.append(i)

    return max(list_index)


result = row_with_min_sum(matrix)
print(result)
```

Bài 5: 
Cho ma trận vuông cấp n. In ra đường chéo chính và đường chéo phụ của ma trận


```python
n = int(input())
matrix = [list(map(int, input().split())) for _ in range(n)]

for i in range(n):
    for j in range(n):
        if i == j:
            print(matrix[i][j], end=' ')
        else:
            print(" ", end=' ')
    print()

print()

for i in range(n):
    for j in range(n):
        if j == n - 1 - i:
            print(matrix[i][j], end=' ')
        else:
            print(" ", end=' ')
    print()
```

Bài 6: Cho ma trận nxm số nguyên, thêm vào một cột tại vị trí k.

```python
n, m = map(int, input().split())
k = int(input())

matrix = [list(map(int, input().split())) for _ in range(n)]
new_col = list(map(int, input().split()))

for i in range(n):
    matrix[i].insert(k-1, new_col[i])

for i in range(n):
    for j in range(m+1):
        print(matrix[i][j], end=' ')
    print()
    


```

Bài 7: Cho ma trận số nguyên nxm, xóa đi hàng k (0 <= k < n).

```python
n, m = map(int, input().split())
k = int(input())
matrix = [list(map(int, input().split())) for _ in range(n)]

for i in range(n):
    if i == k-1:
        matrix.remove(matrix[i])
        break

for i in range(n-1):
    for j in range(m):
        print(matrix[i][j], end=' ')
    print()
```
