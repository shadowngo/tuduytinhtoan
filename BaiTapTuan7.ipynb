W7A1
Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target

```python
def binary_search(arr, left, right, target):
    if left > right:
        return -1  # Target not found
    mid = (left + right) // 2
    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        right = mid - 1
    else:
        left = mid + 1
    return binary_search(arr, left, right, target)

arr = list(map(int, input("Enter sorted array elements separated by space: ").split()))
target = int(input("Enter the target element to search: "))
print(binary_search(arr, 0, len(arr)-1, target))
```

W7A3
Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort); Sắp xếp chọn (Selection
Sort)

```python
def buble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j 
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
         
arr = list(map(int, input("Enter array elements separated by space: ").split()))
print(buble_sort(arr))
```

W7A4
Viết hàm tìm phần tử có số
lần xuất hiện nhiều nhất
trong danh sách. Nếu có
nhiều phần tử cùng số lần,
trả về phần tử xuất hiện
trước.

```python
def find_most_frequent(arr):
    if not arr:
        return None
    
    count_dict = {}
    for item in arr:
        if item in count_dict:
            count_dict[item] += 1
        else:
            count_dict[item] = 1

    most_frequent_item = arr[0] 
    max_count = count_dict[arr[0]]
    
    for item in arr:
        if count_dict[item] > max_count:
            max_count = count_dict[item]
            most_frequent_item = item
            
    return f"{most_frequent_item} xuat hien nhieu nhat, som nhat, {max_count} lan"

arr = list(map(int, input().split()))
print(find_most_frequent(arr))
```

W7A7
Cho một danh sách các số
nguyên nums, hãy tìm độ
dài của dãy con tăng
nghiêm ngặt dài nhất
(không nhất thiết phải liền
kề).

```python
def length_of_lis(nums):
    if not nums:
        return 0
        
    dp = [1] * len(nums)

    for i in range(len(nums)):
        for j in range(i):
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)
                
    return max(dp)
```

W7A8
Bạn được cung cấp: một
danh sách các khoảng thời
gian interval = [[l1, r1], [l2,
r2], ...] và một danh sách
các truy vấn queries = [q1,
q2, ...]
Với mỗi truy vấn q, hãy tìm
độ dài nhỏ nhất của
khoảng thời gian [li, ri] sao
cho li ≤ q ≤ ri.
Nếu không tồn tại khoảng
thời gian nào như vậy, hãy
trả về -1 cho truy vấn đó.
solve in < O(m.n), hint:
binary search in the
queries list

```python
import heapq

def min_interval_duration(intervals, queries):

    intervals.sort(key=lambda x: x[0])
    
    sorted_queries = sorted([(q, i) for i, q in enumerate(queries)])

    res = [-1] * len(queries)
    
    min_heap = []
    i = 0  
    n = len(intervals)

    for q, original_idx in sorted_queries:
        while i < n and intervals[i][0] <= q:
            l, r = intervals[i]
            duration = r - l + 1
            heapq.heappush(min_heap, (duration, r))
            i += 1
            
        while min_heap and min_heap[0][1] < q:
            heapq.heappop(min_heap)
            
        if min_heap:
            res[original_idx] = min_heap[0][0]
            
    return res

intervals = eval(input())
queries = eval(input())

print(min_interval_duration(intervals, queries))
```

W7A9
Cho một danh sách các số
nums, với mỗi phần tử
nums[i], hãy đếm xem có
bao nhiêu phần tử bên phải
nhỏ hơn nums[i].
Trả về danh sách các số
đếm.
Efficient solution: using
heap or binary search tree

```python
import bisect

def count_smaller(nums):
    res = []
    seen = []  

    for num in reversed(nums):
        idx = bisect.bisect_left(seen, num)
 
        res.append(idx)

        bisect.insort(seen, num)
    return res[::-1]

nums = list(map(int, input().split()))
print(count_smaller(nums)) 
```

W7A10
Cho một danh sách các số
nguyên nums và một số
nguyên m, hãy tìm tổng
mảng con lớn nhất theo
modulo m.
Tức là, với bất kỳ mảng
con nums[i:j] nào, hãy tính
sum(nums[i:j]) % m và trả
về kết quả lớn nhất có thể.
Gợi ý:
Duy trì tổng tiền tố (prefix
sum) mod m và sử dụng
tìm kiếm nhị phân để tìm
tổng tiền tố nhỏ nhất chỉ
lớn hơn tổng hiện tại một
chút (để tối đa hóa hiệu số
bao quanh).

```python
import bisect

def max_subarray_sum_modulo(nums, m):
    max_sum = 0
    current_prefix = 0
    seen_prefixes = [0]
    
    for num in nums:
        current_prefix = (current_prefix + num) % m
        max_sum = max(max_sum, current_prefix)
        idx = bisect.bisect_right(seen_prefixes, current_prefix)
        if idx < len(seen_prefixes):
            x = seen_prefixes[idx]
            max_sum = max(max_sum, (current_prefix - x + m) % m)
        bisect.insort(seen_prefixes, current_prefix)
        
    return max_sum

nums = eval(input())
m = int(input())

print(max_subarray_sum_modulo(nums, m))
```
