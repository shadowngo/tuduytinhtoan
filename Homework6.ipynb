Nhậ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k)

```python
n, m, k = map(int, input().split())
matrix = [list(map(int, input().split())) for _ in range(n)]

sum_col = 0
for i in range(n):
    sum_col += matrix[i][k-1]
print(sum_col)
        
sum_row = sum(matrix[k-1])
print(sum_row)

for row in matrix:
    for element in row:
        print(f"{element:4}", end="")
    print()
```

W6A19
Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def index_of_max_in_list(lst):
    max_value = lst[0]
    max_index = 0
    for i in range(1, len(lst)):
        if lst[i] > max_value:
            max_value = lst[i]
            max_index = i
    return max_index

lst = list(map(int, input().split()))
print(lst)
print(index_of_max_in_list(lst))
```

W6A20
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def index_of_k_in_list(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i
    return -1

lst = list(map(int, input().split()))
k = int(input())
print(lst)
print(index_of_k_in_list(lst, k))
```

W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
string = list(map(str, input().split()))

dct = {}

for index, word in enumerate(string):
    if word not in dct:
        dct[word] = []
    dct[word].append(index)

for word in dct:
    dct[word] = tuple(dct[word])

print(dct)
```

W6A24
Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9

```python
lst = list(map(int, input().split()))
zero_lst = []
nonzero_lst = []

for x in lst:
    if x == 0:
        zero_lst.append(x)
    else:
        nonzero_lst.append(x)
print(*zero_lst + nonzero_lst)
```

W6A26
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối

```python
lst = list(map(int, input().split()))
sum_peaks = 0

for i in range(1, len(lst)-1):
    if lst[i-1] < lst[i] > lst[i+1]:
        sum_peaks += lst[i]
print(sum_peaks)
```

W6A27
Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).
nếu không có in ra -1

```python
arr = list(map(int, input().split()))
index = []
x = int(input())

for i, element in enumerate(arr):
    if element < x:
        index.append(i)

if index == []:
    print(-1)

print(*index)
```

W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
lst = list(map(int, input().split()))

def tang_giam(lst):
    for i in range(1, len(lst)-1):
        if lst[i-1] < lst[i] < lst[i+1]:
            return "tang"
        if lst[i-1] > lst[i] < lst[i+1]:
            return "giam"
    return "vo danh"
print(tang_giam(lst))

```
