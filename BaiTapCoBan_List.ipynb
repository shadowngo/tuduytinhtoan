### Bài 3: Tìm giá trị trung bình của các số nguyên dương trong list

```python
lst = list(map(int, input().split()))
s = 0
count = 0

for i in lst:
    if i > 0:
        s += i
        count += 1

if count > 0:
    print(round((s / count), 2))

```

### Bài 5: Đưa tất cả các số bằng 0 trong list về đầu của list. Các số khác không thay đổi thứ tự trước sau
3 4 6 7 1 9 0 0 23 0 0 34 67 89
0 0 0 0 3 4 6 7 1 9 23 34 67 89

```python
lst = list(map(int, input().split()))
lst0 = [x for x in lst if x == 0]
lst_non_zero = [x for x in lst if x != 0]

new_list = lst0 + lst_non_zero
print(*new_list)
```

### Bài 6: Xóa tất cả các số giống nhau trong list, Các số khác giữ nguyên vị trí trước sau

```python
numbers = list(map(int, input().split()))
giong_nhau = []

for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] == numbers[j]:
            giong_nhau.append(numbers[i])

khac_nhau = [x for x in numbers if x not in giong_nhau]
print(*khac_nhau)

```

### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang
9 8 7 6 5 4 0: Giam
1 2 3 7 1 0 2: Vo danh
1 2 2 4 5 7: Vo danh

```python
lst = list(map(int, input().split()))

def lst_tang(lst):
    for i in range(len(lst) - 1):
        if lst[i] >= lst[i + 1]:
            return False
    return True

def lst_giam(lst):
    for i in range(len(lst) - 1):
        if lst[i] <= lst[i + 1]:
            return False
    return True

if lst_tang(lst):
    print("Tang")
elif lst_giam(lst):
    print("Giam")
else:
    print("Vo danh")
   
```

### Bài 8: Kiểm tra một list số nguyên có phải là ziczac khong
1 3 2 5 2 8 7 9 2 100: ziczac
34 2 56 7 90 20 30 7 120: ziczac

```python
lst = list(map(int, input().split()))

def ziczac_1(lst):
    for i in range(len(lst) - 1):
        if i % 2 == 0:
            if lst[i] >= lst[i + 1]:
                return False
        else:
            if lst[i] <= lst[i + 1]:
                return False
    return True

def ziczac2(lst):
    for i in range(len(lst) - 1):
        if i % 2 != 0:
            if lst[i] >= lst[i + 1]:
                return False
        else:
            if lst[i] <= lst[i + 1]:
                return False
    return True

if ziczac_1(lst):
    print("Zic zac")
if ziczac2(lst):
    print("Zic zac")
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước. Nếu không có k trong ds thì in ra "-1".
VD: k = 3, [1,3,4,4,3,3,4,3,6,1] => k xuất hiện tại các vị trí (trả về 1 list) [1, 4, 5, 7]
k = 10, [1,3,4,4,3,3,4,3,6,1] => k không xuất hiện trong ds => trả về "-1"

```python
k = int(input())
lst = list(map(int, input().split()))

for i, x in enumerate(lst):
    if k == x:
        print(i, end=' ')
if k not in lst:
    print(-1)


    
```
